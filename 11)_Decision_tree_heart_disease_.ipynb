Interview questions
1) What are some common hyperparameters of Decision Tree models, and how do they affect the model's performance?

Some common Decision Tree hyperparameters are:

max_depth – Controls the maximum depth of the tree. A smaller value prevents overfitting, while a very large value can make the tree overfit.
min_samples_split – Specifies the minimum number of samples required to split an internal node. Increasing it can reduce overfitting.
min_samples_leaf – Specifies the minimum number of samples that must be present in a leaf node. Higher values make the tree simpler and more stable.
criterion – Determines how the quality of a split is measured. Common options are Gini impurity and Entropy.
max_features – Determines the number of features considered when finding the best split.

2) What is the difference between Label Encoding and One-Hot Encoding?

Label Encoding	One-Hot Encoding
Converts categories into numerical labels.	Creates a separate binary column for each category.
Example: Male = 0, Female = 1.	Example: Male → [1,0], Female → [0,1].
Uses a single column.	Creates multiple columns.
Can sometimes imply an order between categories.	Does not imply any order.
