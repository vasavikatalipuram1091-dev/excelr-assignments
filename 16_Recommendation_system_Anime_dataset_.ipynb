```python
# Check unique values in important categorical columns

print("Types:")
print(df['type'].unique())

print("\nNumber of unique genres:")
print(df['genre'].nunique())
```

```python
# Handle missing values

df['genre'] = df['genre'].fillna('Unknown')
df['type'] = df['type'].fillna('Unknown')
df['rating'] = df['rating'].fillna(df['rating'].median())

```

```python
# Check again
print(df.isnull().sum())
```

```python
df.info()
```

```python
# Convert episodes to numeric
# Unknown episode counts will become NaN

df['episodes'] = pd.to_numeric(df['episodes'], errors='coerce')

# Fill missing episode values with the median
df['episodes'] = df['episodes'].fillna(df['episodes'].median())

print(df['episodes'].dtype)
print(df['episodes'].isnull().sum())
```

```python
df[['episodes', 'rating', 'members']].describe()
```

```python
# One-hot encode anime type

type_encoded = pd.get_dummies(
    df['type'],
    prefix='type'
)

type_encoded.head()
```

```python
# Create genre features

genre_features = df['genre'].str.get_dummies(sep=', ')

genre_features.head()
```

```python
# Combine genre and type features with numerical features

features = pd.concat(
    [
        genre_features,
        type_encoded,
        df[['episodes', 'rating', 'members']]
    ],
    axis=1
)

print("Feature matrix shape:", features.shape)
features.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

features_scaled = scaler.fit_transform(features)

print("Scaled feature shape:", features_scaled.shape)
```

```python
features_scaled[:5]
```

```python
#Calculate Cosine Similarity
from sklearn.metrics.pairwise import cosine_similarity

# Calculate cosine similarity between all anime
cosine_sim = cosine_similarity(features_scaled)

print("Cosine Similarity Matrix Shape:", cosine_sim.shape)
```

```python
# Create anime recommendation function

def recommend_anime(anime_title, num_recommendations=10):

    # Find the index of the selected anime
    matches = df[df['name'].str.lower() == anime_title.lower()]

    if matches.empty:
        print("Anime not found.")
        return

    index = matches.index[0]

    # Get similarity scores for the selected anime
    similarity_scores = list(enumerate(cosine_sim[index]))

    # Sort by similarity score from highest to lowest
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove the selected anime itself
    similarity_scores = [
        item for item in similarity_scores
        if item[0] != index
    ]

    # Get top recommendations
    top_recommendations = similarity_scores[:num_recommendations]

    recommendations = []

    for i, score in top_recommendations:
        recommendations.append({
            'Anime': df.iloc[i]['name'],
            'Similarity Score': score
        })

    return pd.DataFrame(recommendations)
```

```python
recommend_anime('Naruto', 10)
```

```python
# Recommendation using a similarity threshold

def recommend_by_threshold(anime_title, threshold=0.5):

    matches = df[df['name'].str.lower() == anime_title.lower()]

    if matches.empty:
        print("Anime not found.")
        return

    index = matches.index[0]

    similarity_scores = list(enumerate(cosine_sim[index]))

    recommendations = []

    for i, score in similarity_scores:
        if i != index and score >= threshold:
            recommendations.append({
                'Anime': df.iloc[i]['name'],
                'Similarity Score': score
            })

    recommendations = sorted(
        recommendations,
        key=lambda x: x['Similarity Score'],
        reverse=True
    )

    return pd.DataFrame(recommendations)
```

```python
print("Threshold = 0.3")
print(recommend_by_threshold('Naruto', 0.3).head(10))
```

```python
print("Threshold = 0.5")
print(recommend_by_threshold('Naruto', 0.5).head(10))
```

```python
print("Threshold = 0.7")
print(recommend_by_threshold('Naruto', 0.7).head(10))
```

1. Difference between User-Based and Item-Based Collaborative Filtering

User-based collaborative filtering recommends items by finding users who have similar preferences to the current user. Then it recommends items that those similar users liked.

Item-based collaborative filtering recommends items by finding items that are similar to the items the user has already liked or interacted with.
