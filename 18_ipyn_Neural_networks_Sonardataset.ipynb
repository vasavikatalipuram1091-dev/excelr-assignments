```python
X = df.drop('Y', axis=1)
y = df['Y']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
y = y.map({'M': 1, 'R': 0})

y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Build the ANN model
ann_model = Sequential([
    Dense(16, activation='relu', input_shape=(60,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

```

```python
# Display model architecture
ann_model.summary()
```

```python
ann_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = ann_model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss, test_accuracy = ann_model.evaluate(X_test, y_test, verbose=0)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
y_pred_prob = ann_model.predict(X_test)

y_pred = (y_pred_prob > 0.5).astype(int).ravel()

print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-Score :", f1)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Rock', 'Mine']
).plot()

plt.title("ANN Confusion Matrix")
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

def create_ann(hidden_layers=1, neurons=16, activation='relu', learning_rate=0.001):

    model = Sequential()

    # First hidden layer
    model.add(Dense(neurons, activation=activation, input_shape=(60,)))

    # Additional hidden layers
    for _ in range(hidden_layers - 1):
        model.add(Dense(neurons, activation=activation))

    # Output layer
    model.add(Dense(1, activation='sigmoid'))

    # Compile
    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
configs = [
    {'hidden_layers': 1, 'neurons': 8,  'activation': 'relu', 'learning_rate': 0.001},
    {'hidden_layers': 1, 'neurons': 16, 'activation': 'relu', 'learning_rate': 0.001},
    {'hidden_layers': 2, 'neurons': 16, 'activation': 'relu', 'learning_rate': 0.001},
    {'hidden_layers': 2, 'neurons': 32, 'activation': 'relu', 'learning_rate': 0.001},
    {'hidden_layers': 2, 'neurons': 16, 'activation': 'tanh', 'learning_rate': 0.001},
    {'hidden_layers': 2, 'neurons': 16, 'activation': 'relu', 'learning_rate': 0.01}
]

results = []

for config in configs:

    model = create_ann(**config)

    model.fit(
        X_train,
        y_train,
        epochs=50,
        batch_size=16,
        validation_split=0.2,
        verbose=0
    )

    _, accuracy = model.evaluate(X_test, y_test, verbose=0)

    results.append({
        **config,
        'test_accuracy': accuracy
    })

results
```

```python
results_df = pd.DataFrame(results)

results_df
```

```python
best_result = results_df.loc[results_df['test_accuracy'].idxmax()]

print("Best Hyperparameters:")
print(best_result)
```

```python
best_model = create_ann(
    hidden_layers=int(best_result['hidden_layers']),
    neurons=int(best_result['neurons']),
    activation=best_result['activation'],
    learning_rate=float(best_result['learning_rate'])
)

best_model.summary()
```

```python
best_history = best_model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
tuned_loss, tuned_accuracy = best_model.evaluate(
    X_test, y_test, verbose=0
)

print("Tuned ANN Test Loss:", tuned_loss)
print("Tuned ANN Test Accuracy:", tuned_accuracy)
```

```python
y_tuned_prob = best_model.predict(X_test)

y_tuned_pred = (y_tuned_prob > 0.5).astype(int).ravel()
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

tuned_accuracy = accuracy_score(y_test, y_tuned_pred)
tuned_precision = precision_score(y_test, y_tuned_pred)
tuned_recall = recall_score(y_test, y_tuned_pred)
tuned_f1 = f1_score(y_test, y_tuned_pred)

print("Tuned ANN Results")
print("-------------------")
print("Accuracy :", tuned_accuracy)
print("Precision:", tuned_precision)
print("Recall   :", tuned_recall)
print("F1-Score :", tuned_f1)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_tuned = confusion_matrix(y_test, y_tuned_pred)

print("Tuned ANN Confusion Matrix:")
print(cm_tuned)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tuned,
    display_labels=['Rock', 'Mine']
).plot()

plt.title("Tuned ANN Confusion Matrix")
plt.show()
```

```python
comparison = pd.DataFrame({
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score'],
    'Basic ANN': [
        accuracy,
        precision,
        recall,
        f1
    ],
    'Tuned ANN': [
        tuned_accuracy,
        tuned_precision,
        tuned_recall,
        tuned_f1
    ]
})

comparison
```

Discussion

The Sonar dataset contains 208 observations with 60 numerical features representing the energy of sonar signals across different frequency bands. The target variable contains two classes: Mine (M) and Rock (R). The dataset was checked for missing values and duplicate records, and the features were standardized before training the Artificial Neural Network (ANN).

A basic ANN model was initially developed using hidden layers with ReLU activation and a sigmoid output layer. The basic model achieved an accuracy of 90.48%, with a precision of 87.50%, recall of 95.45%, and F1-score of 91.30%.

Conclusion

The ANN successfully classified sonar signals into Mine and Rock categories. Data preprocessing, feature standardization, model training, and systematic hyperparameter tuning were performed as required.
