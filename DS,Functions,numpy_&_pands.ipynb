```python
# 1. Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

# Create a list of 5 integers
my_list = [10, 20, 30, 40, 50]
print(f"Original list: {my_list}")

# append: Add an element to the end of the list
my_list.append(60)
print(f"After append(60): {my_list}")

# extend: Add elements of an iterable (e.g., another list) to the end of the list
my_list.extend([70, 80])
print(f"After extend([70, 80]): {my_list}")

# insert: Add an element at a specific index
my_list.insert(1, 15)
print(f"After insert(1, 15): {my_list}")

# remove: Remove the first occurrence of a specified value
my_list.remove(30)
print(f"After remove(30): {my_list}")

# pop: Remove and return the element at a specified index (or the last element if no index is given)
popped_element = my_list.pop(0) # Remove element at index 0
print(f"After pop(0): {my_list}, Popped element: {popped_element}")

# index: Return the index of the first occurrence of a specified value
index_of_40 = my_list.index(40)
print(f"Index of 40: {index_of_40}")

# count: Return the number of times a specified value appears in the list
my_list.append(40) # Add another 40 for demonstration
count_of_40 = my_list.count(40)
print(f"After appending another 40, count of 40: {count_of_40}")

# sort: Sort the list in ascending order (in-place)
my_list.sort()
print(f"After sort(): {my_list}")

# reverse: Reverse the order of elements in the list (in-place)
my_list.reverse()
print(f"After reverse(): {my_list}")

# clear: Remove all elements from the list
my_list.clear()
print(f"After clear(): {my_list}")
```
