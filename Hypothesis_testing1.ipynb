```python
import pandas as pd
import numpy as np
from scipy import stats

# Durability data
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85,
        1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

print('Sample size:', n)
print('Sample mean:', round(mean, 4))
print('Sample standard deviation:', round(s, 4))

# Part (a): 99% CI using t-distribution
t_value = stats.t.ppf(0.995, df=n-1)
margin_t = t_value * s / np.sqrt(n)
ci_t = (mean - margin_t, mean + margin_t)

print('\n99% CI using sample SD (t-distribution):')
print(ci_t)

# Part (b): 99% CI using known population SD
sigma = 0.2
z_value = stats.norm.ppf(0.995)
margin_z = z_value * sigma / np.sqrt(n)
ci_z = (mean - margin_z, mean + margin_z)

print('\n99% CI using known population SD (z-distribution):')
print(ci_z)
```
