```python
import pandas as pd
import numpy as np

# Step 1: Create a DataFrame with 10 rows
data = {
    'Name': ['Asha', 'Ravi', 'Sita', 'Kiran', 'Mohan',
             'Divya', 'Ravi', 'Sneha', 'Arjun', 'Pooja'],
    'Age': [23, 25, 22, 28, 30, 27, 25, 24, 29, 26],
    'City': ['Hyderabad', 'Chennai', 'Delhi', 'Mumbai', 'Pune',
             'Hyderabad', 'Chennai', 'Bangalore', 'Delhi', 'Pune'],
    'Salary': [45000, 52000, np.nan, 60000, 58000,
               62000, 52000, 48000, np.nan, 55000]
}

df = pd.DataFrame(data)

print('Original DataFrame:')
print(df)

# Add a duplicate row for demonstration
df = pd.concat([df, df.iloc[[1]]], ignore_index=True)

print('\nDataFrame after adding a duplicate row:')
print(df)

# Step 2: Use .info()
print('\nDataFrame Info:')
print(df.info())

# Step 3: Use .describe()
print('\nDataFrame Description:')
print(df.describe())

# Step 4: Select 'Name' and 'City' columns
print('\nSelected Columns (Name and City):')
print(df[['Name', 'City']])

# Step 5: Drop 'City' column
df_no_city = df.drop('City', axis=1)

print('\nDataFrame after dropping City column:')
print(df_no_city)

# Step 6: Fill missing Salary values with the mean
mean_salary = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean_salary)

print('\nDataFrame after filling missing Salary values:')
print(df)

# Step 7: Remove duplicate rows
df_clean = df.drop_duplicates()

print('\nDataFrame after removing duplicates:')
print(df_clean)
```
