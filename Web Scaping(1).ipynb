```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.flipkart.com/search?q=mobiles+under+25000&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page=1"

r = requests.get(url)
print(r)
```

```python
soup = BeautifulSoup(r.text,"lxml")
soup.prettify()  # it is a function  used to format HTML/XML code in a clean, readable format
```

```python
phone_name = []
for phone in soup.find_all("div", class_="KzDlHZ"):
    phone_name.append(phone.text.strip())
```

```python
phone_name
```

```python
phone_price = []

for phone in soup.find_all("div", class_="Nx9bqj _4b5DiR"):
    phone_price.append(phone.text.strip())
```

```python
phone_price
```

```python
df = pd.DataFrame({
    "Names" : phone_name,
    "Prices": phone_price
})
```

```python
df
```

```python
df.to_csv("phones.csv", index=False)
```
