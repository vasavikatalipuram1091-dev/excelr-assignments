```python
num = int(input("Enter a number:7"))
if num <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")

```

```python
#Exercise 2: Product of Random Numbers

#Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.
```

```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

answer = int(input(f"What is {num1} × {num2}? "))

product = num1 * num2

if answer == product:
    print("Correct! Well done.")
else:
    print("Incorrect.")

    print("The correct answer is:", product)
```

```python
# Hospital Billing System

room_charge_per_day = 2000
doctor_fee = 1500
lab_test_charge = 300
days = 4
lab_tests = 3
medicine_charge = 2400
room_charge = room_charge_per_day * days
lab_charge = lab_tests * lab_test_charge
total_bill = room_charge + doctor_fee + lab_charge + medicine_charge

if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

print("Room Charges:", room_charge)
print("Doctor Fee:", doctor_fee)
print("Lab Test Charges:", lab_charge)
print("Medicine Charges:", medicine_charge)
print("Total Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
```
